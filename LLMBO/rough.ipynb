```python
from langchain.prompts import FewShotPromptTemplate, PromptTemplate

# Define a list of examples
examples = [
    {"input": "What is the capital of France?", "output": "The capital of France is Paris."},
    {"input": "What is the capital of Spain?", "output": "The capital of Spain is Madrid."}
]

# Define the example template with placeholders
example_prompt = PromptTemplate(
    input_variables=["input", "output"],
    template="Input: {input} Output: {output}"
)

# Create a FewShotPromptTemplate object with the examples and the example template
few_shot_prompt = FewShotPromptTemplate(
    examples=examples,
    example_prompt=example_prompt,
    prefix="Here are some examples of question and answer pairs:\n",
    suffix="\nGiven the above examples, answer the following question.\nInput: {input}\nOutput:",
    input_variables=["input"]
)
```

```python
# Define the new input for which you want a prediction
new_input = "What is the capital of Italy?"

# Generate the prompt by filling in the new input
filled_prompt = few_shot_prompt.format(input=new_input)

print(filled_prompt)
```
