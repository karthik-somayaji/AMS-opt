```python
import numpy as np

def fn(p, v, a1, a2):
     prange = [30e-6, 1.5e-3]
     x1 = -1* (np.log(p) -np.log(prange[0]) )/ (np.log(prange[1]) - np.log(prange[0]))
     x2 = -v

     return a1*x1 + a2*x2

a1, a2 = 0.75, 0.25
print(fn(15.25e-6, 0.091, a1, a2))
print(fn(16.57e-6, 0.014, a1, a2))
print(fn(13.13e-6, 0.006, a1, a2))
print(fn(14.26e-6, 0.003, a1, a2))
```
