```python
import torch
from botorch.models import SingleTaskGP
from botorch.fit import fit_gpytorch_model
from botorch.acquisition import ExpectedImprovement
from gpytorch.mlls import ExactMarginalLogLikelihood

# Example data
train_X = torch.rand(10, 1) * 5  # 10 data points, 1D, domain [0, 5]
train_Y = torch.sin(train_X) + 0.2 * torch.randn_like(train_X)  # noisy sine wave

# Define and fit GP model
gp_model = SingleTaskGP(train_X, train_Y)
mll = ExactMarginalLogLikelihood(gp_model.likelihood, gp_model)
fit_gpytorch_model(mll)

# Setup acquisition function
best_f = train_Y.max()  # best observed value (for EI)
ei = ExpectedImprovement(model=gp_model, best_f=best_f)

# Define multiple points 'x' where you want to evaluate the acquisition function
x = torch.tensor([[1.0], [2.0], [3.0], [4.0]]).unsqueeze(2) # Shape: (4, 1)

# Evaluate the acquisition function at these points
acquisition_values = ei(x)

print("Acquisition values at multiple points:", acquisition_values)
```
